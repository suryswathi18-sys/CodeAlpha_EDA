## EDA Findings — Retail Sales Dataset

**Dataset Overview:**
- 1000 transactions, 9 columns, no missing values
- Categories: Clothing, Electronics, Beauty (fairly balanced)
- Gender split: 510 Female, 490 Male (balanced)

**Key Insights:**

1. **Electronics generates the highest revenue** (₹156,905) despite having fewer transactions than Clothing, and having similar average price-per-unit and quantity-per-transaction to other categories — suggesting revenue is driven by a combination of small factors rather than one dominant variable.

2. **Price-per-unit and quantity purchased are consistent across all product categories** (~₹175-184 avg price, ~2.5 avg quantity), showing customer purchasing behavior doesn't vary much by category.

3. **Age 46-55 is the top-spending demographic** (₹100,690 total), followed by 26-35 (₹98,480). Younger (18-25) and older (56-65) customers spend the least — consistent with peak-earning-age spending patterns.

4. **Total Amount has high variability** (std ₹560 vs mean ₹456), meaning spending per transaction is quite spread out rather than clustered near the average — some customers make small purchases, others large ones.

**Conclusion:** No single factor (price, quantity, gender) strongly drives category-level revenue differences — Electronics' lead appears to come from a combination of small advantages. Age is the clearest differentiator in customer spending behavior.
